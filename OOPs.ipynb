Ques.1 Create student class that takes name and makes of 3 subjects as arguments in constructor. Then create method to print the average.

```python
class student():
    def __init__(self, name, marks):
        self.name = name 
        self.marks = marks 

    def get_avg(self):
        sum = 0
        for i in self.marks:
            sum+=i
        print("Hi",self.name, "your average score is",sum/3)

s1 = student("karan,",[23,45,67])
s1.get_avg()
```

Ques.2 Create an account class with 2 attributes- balance and account no. 
       Create methods for debit,credit and printing the balance 

```python
class account():
    def __init__(self, bal,act):
        self.balance = bal
        self.acc= act

    def debit(self,value):
        self.balance -= value
        print("Rs.",value,"is debited from your account No.",self.acc)
        print("Total balamce =",self.balance)  

    
    def credit(self,value):
        self.balance += value
        print("Rs.",value,"is credited in your account No.",self.acc)
        print("Total balamce =",self.balance)    

user1 = account(40000, 12345)
user1.debit(1000)
user1.credit(100)
```
